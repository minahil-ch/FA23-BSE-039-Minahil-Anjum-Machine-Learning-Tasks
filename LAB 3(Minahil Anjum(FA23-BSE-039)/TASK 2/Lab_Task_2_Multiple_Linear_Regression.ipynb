# Lab Task 2: Multiple Linear Regression
**Goal:** Using `hiring.csv` (experience, test score, interview score -> salary),
build a model for HR to decide salaries for future candidates, then predict salary for:
1. 2 yr experience, 9 test score, 6 interview score
2. 12 yr experience, 10 test score, 10 interview score

This is **Multiple** Linear Regression because we now use 3 features instead of 1.

```
import numpy as np
import pandas as pd
import math
from sklearn.linear_model import LinearRegression
```

## Step 1: Upload and load the dataset

```
from google.colab import files
uploaded = files.upload()   # choose hiring.csv
```

```
df = pd.read_csv("hiring.csv")
df
```

## Step 2: Clean the data
Real data is messy. This dataset has 2 problems:
1. `experience` column has some blank cells (meaning 0 years) AND some values are
   written as words ("five", "two", "ten"...) instead of numbers.
2. `test_score(out of 10)` has one missing value (NaN).

We fix both issues below.

```
# pip install the word2number package (converts "five" -> 5, "eleven" -> 11, etc.)
!pip install word2number -q
from word2number import w2n
```

```
# 2a. Fill blank experience cells with 'zero' (blank means the candidate had no prior experience)
df['experience'] = df['experience'].fillna('zero')

# 2b. Convert every experience value (word or NaN-filled 'zero') into an integer
df['experience'] = df['experience'].apply(w2n.word_to_num)

df
```

```
# 2c. Fill the missing test score with the column's MEDIAN
#     (median is safer than mean here because we only have 8 rows and no outliers to worry about,
#      but either is acceptable for this small exercise)
median_test_score = math.floor(df['test_score(out of 10)'].median())
print("Median test score used to fill missing value:", median_test_score)

df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(median_test_score)
df
```

## Step 3: Prepare X (features) and y (target)

```
X = df[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']]
y = df['salary($)']
```

## Step 4: Train the Multiple Linear Regression model
With only 8 rows of data, we train on the entire dataset (no train/test split) --
the point of this exercise is to learn the relationship, not to evaluate generalisation.

```
regressor = LinearRegression()
regressor.fit(X, y)
```

## Step 5: Predict salaries for the two given candidates

```
# Candidate 1: 2 yr experience, 9 test score, 6 interview score
candidate_1 = regressor.predict([[2, 9, 6]])
print("Predicted salary for candidate 1 (2 yr, 9, 6):", round(candidate_1[0], 2))

# Candidate 2: 12 yr experience, 10 test score, 10 interview score
candidate_2 = regressor.predict([[12, 10, 10]])
print("Predicted salary for candidate 2 (12 yr, 10, 10):", round(candidate_2[0], 2))
```

## Step 6: Inspect the model coefficients (optional, for understanding)
`salary = c1*experience + c2*test_score + c3*interview_score + intercept`

```
print("Coefficients [experience, test_score, interview_score]:", regressor.coef_)
print("Intercept:", regressor.intercept_)
```

## Conclusion
The model learned how experience, test score, and interview score jointly influence salary,
and used that relationship to predict fair salaries for two new candidates.
