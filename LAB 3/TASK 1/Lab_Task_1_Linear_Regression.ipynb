# Lab Task 1: Linear Regression
**Goal:** Predict Canada's per capita income in year 2020 using `canada_per_capita_income.csv`.

We only have ONE feature (`year`), so this is a **Simple Linear Regression** problem:
we fit a straight line `income = m * year + c` and use it to predict income for 2020.

```
# Step 1: Import the libraries we need
import numpy as np              # for numeric arrays
import pandas as pd             # for reading/handling the csv as a dataframe
import matplotlib.pyplot as plt # for plotting the data and the regression line
from sklearn.linear_model import LinearRegression   # the model we will train
```

## Step 2: Load the dataset
Upload `canada_per_capita_income.csv` when prompted (it is included in this zip file).

```
# Read the csv into a pandas dataframe
df = pd.read_csv("canada_per_capita_income.csv")
df.head()   # show first 5 rows to check it loaded correctly
```

## Step 3: Visualise the data
Always look at your data before modelling it -- a scatter plot tells us if a straight
line is actually a reasonable fit.

```
plt.xlabel('Year')
plt.ylabel('Per Capita Income (US$)')
plt.title('Canada Per Capita Income over the years')
plt.scatter(df['year'], df['per capita income (US$)'], color='red', marker='+')
plt.show()
```

## Step 4: Prepare X (feature) and y (target)
`LinearRegression` in sklearn expects X to be **2-D** (a table of features), so we use
`df[['year']]` (double brackets) instead of `df['year']` (which would be 1-D).

```
X = df[['year']]                       # feature: year  (2-D dataframe)
y = df['per capita income (US$)']      # target: income (1-D series)
```

## Step 5: Train the Linear Regression model
The dataset is small (only 47 rows spanning 1970-2016) so, unlike the earlier Salary
example, here we train on the FULL dataset instead of splitting into train/test --
our real goal is to extrapolate to year 2020, not to test on held-out historical years.

```
regressor = LinearRegression()
regressor.fit(X, y)   # this finds the best-fit slope (m) and intercept (c)
```

## Step 6: Plot the regression line on top of the data
This lets us visually confirm the line is a good fit.

```
plt.xlabel('Year')
plt.ylabel('Per Capita Income (US$)')
plt.title('Canada Per Capita Income - Best Fit Line')
plt.scatter(df['year'], df['per capita income (US$)'], color='red', marker='+', label='Actual data')
plt.plot(df['year'], regressor.predict(X), color='blue', label='Best fit line')
plt.legend()
plt.show()
```

## Step 7: Predict per capita income for year 2020

```
prediction_2020 = regressor.predict([[2020]])
print("Predicted per capita income for Canada in 2020: $", round(prediction_2020[0], 2))
```

## Step 8: Inspect the line equation (optional, for understanding)
`income = coefficient * year + intercept`

```
print("Coefficient (slope, m):", regressor.coef_[0])
print("Intercept (c):", regressor.intercept_)

# Manual verification of the same 2020 prediction using the formula y = m*x + c
manual_2020 = regressor.coef_[0] * 2020 + regressor.intercept_
print("Manual calculation check:", round(manual_2020, 2))
```

## Conclusion
The model predicts Canada's per capita income for 2020 based on the linear trend of the
1970-2016 data. Note: extrapolating 4 years beyond the last data point (2016) assumes the
trend continues linearly, which is a simplifying assumption worth mentioning in your report.
