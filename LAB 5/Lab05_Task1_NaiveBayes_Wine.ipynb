# Lab 05 – Task 1: Wine dataset (Gaussian vs Multinomial Naive Bayes)

```
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.datasets import load_wine
from sklearn.naive_bayes import GaussianNB, MultinomialNB
```

```
wine = load_wine()
X, y = wine.data, wine.target
print(X.shape, wine.target_names)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```
gnb = GaussianNB().fit(X_train, y_train)
mnb = MultinomialNB().fit(X_train, y_train)   # features are non-negative, so this works
acc_g = accuracy_score(y_test, gnb.predict(X_test))
acc_m = accuracy_score(y_test, mnb.predict(X_test))
print(f'Gaussian NB    accuracy: {acc_g:.4f}')
print(f'Multinomial NB accuracy: {acc_m:.4f}')
print('Better model:', 'Gaussian' if acc_g > acc_m else 'Multinomial')
```

```
best = gnb if acc_g >= acc_m else mnb
y_pred = best.predict(X_test)
print(classification_report(y_test, y_pred, target_names=wine.target_names))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted'); plt.ylabel('Truth'); plt.show()
```

```
# Predictions on some test samples
for i in range(5):
    print('Actual:', wine.target_names[y_test[i]], '| Predicted:', wine.target_names[y_pred[i]])
```

**Conclusion:** Wine features are continuous, so Gaussian NB usually performs better than Multinomial NB (which is meant for count/frequency data).
