# Lab 05 – Task 2: Loan dataset – predict customers who have not fully paid
Dataset: `loan_data.csv` (Lending Club data from Kaggle, target column `not.full.paid`). Place it in the same folder as this notebook (or upload in Colab).

```
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler
```

```
df = pd.read_csv('loan_data.csv')
print(df.shape)
df.head()
```

### Data exploration

```
df.info()
print(df.isnull().sum())
print(df['not.full.paid'].value_counts())
df.describe()
```

```
plt.figure(figsize=(6,4)); sns.countplot(x='not.full.paid', data=df); plt.show()
plt.figure(figsize=(10,5)); sns.countplot(x='purpose', hue='not.full.paid', data=df); plt.xticks(rotation=45); plt.show()
plt.figure(figsize=(11,8)); sns.heatmap(df.drop(columns='purpose').corr(), annot=True, fmt='.2f', cmap='coolwarm'); plt.show()
```

### Preprocessing and model

```
data = pd.get_dummies(df, columns=['purpose'], drop_first=True)
X = data.drop(columns='not.full.paid'); y = data['not.full.paid']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101, stratify=y)
sc = StandardScaler().fit(X_train)
Xtr, Xte = sc.transform(X_train), sc.transform(X_test)

nb = GaussianNB().fit(Xtr, y_train)
y_pred = nb.predict(Xte)
print('Accuracy:', accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues'); plt.xlabel('Predicted'); plt.ylabel('Truth'); plt.show()
```

**Note:** the data is imbalanced (most customers fully paid), so look at precision/recall of class 1, not just accuracy.
